```python
import pandas as pd
import openpyxl
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_excel("populacao.xlsx")

df.head()
```

```python
df.columns
```

```python
pop_estado = (
    df.groupby(["COD. UF", "UF"], as_index=False)
      .agg(
          populacao_2010=("POPULAÇÃO 2010", "sum"),
          populacao_2022=("POPULAÇÃO 2022", "sum")
      )
)

pop_estado
```

```python
pop_estado["crescimento"] = (
    pop_estado["populacao_2022"] -
    pop_estado["populacao_2010"]
)

pop_estado
```

```python
pop_estado = pop_estado.sort_values(
    "crescimento",
    ascending=False
)

pop_estado
```

```python
pop_estado.to_csv(
    "populacao_estado.csv",
    sep=";",
    index=False
)
```

```python
pop_municipio = (
    df.groupby("MUNICÍPIO", as_index=False)
      .agg(
          populacao_2010=("POPULAÇÃO 2010", "sum"),
          populacao_2022=("POPULAÇÃO 2022", "sum")
      )
)

pop_municipio
```

```python
pop_municipio["crescimento"] = (
    pop_municipio["populacao_2022"] -
    pop_municipio["populacao_2010"]
)

pop_municipio
```

```python
pop_municipio = pop_municipio.sort_values(
    "crescimento",
    ascending=False
)

pop_municipio
```

```python
pop_municipio.to_csv(
    "populacao_municipio.csv",
    sep=";",
    index=False
)
```

```python
top = pop_estado.head(10)

top = top.melt(
    id_vars="UF",
    value_vars=["populacao_2010", "populacao_2022"],
    var_name="ano",
    value_name="populacao"
)

plt.figure(figsize=(10, 5))

sns.barplot(
    data=top,
    x="UF",
    y="populacao",
    hue="ano"
)

plt.title("10 UFs com maior crescimento: 2010 vs 2022")
plt.ylabel("Habitantes")
plt.xlabel("Estado")
plt.xticks(rotation=30)

plt.tight_layout()
plt.show()
```
